## 1. Inspecionar estrutura HDF5

Antes de implementar `data_loader.py`, é necessário mapear a estrutura interna dos arquivos.

```python
DATA_DIR = '../data/'  # ajuste conforme localização dos arquivos VeHIF

import os
files = [f for f in os.listdir(DATA_DIR) if f.endswith(('.h5', '.hdf5'))]
print(f'{len(files)} arquivos encontrados em {DATA_DIR}')
print('Primeiros 5:', files[:5])
```

```python
# Inspecionar o primeiro arquivo
sample_path = os.path.join(DATA_DIR, files[0])

def print_hdf5_structure(name, obj):
    print(name, '->', type(obj).__name__, getattr(obj, 'shape', ''), getattr(obj, 'dtype', ''))

with h5py.File(sample_path, 'r') as f:
    print('Keys no nível raiz:', list(f.keys()))
    print('\nEstrutura completa:')
    f.visititems(print_hdf5_structure)
    print('\nAtributos do arquivo:', dict(f.attrs))
```

## 2. Visualizar sinais LF e HF

Após identificar as keys corretas acima, ajustar os nomes abaixo.

```python
# TODO: ajustar LF_KEY e HF_KEY conforme estrutura real do HDF5
LF_KEY = 'lf_current'  # placeholder — confirmar após inspeção
HF_KEY = 'hf_current'  # placeholder
FS_LF  = None          # Hz — preencher após inspecionar metadados
FS_HF  = None          # Hz

with h5py.File(sample_path, 'r') as f:
    lf = np.array(f[LF_KEY])
    hf = np.array(f[HF_KEY])

fig, axes = plt.subplots(2, 1, figsize=(14, 6), sharex=False)
axes[0].plot(lf)
axes[0].set_title('Canal LF (corrente)')
axes[0].set_ylabel('Amplitude')
axes[1].plot(hf, color='C1')
axes[1].set_title('Canal HF (corrente)')
axes[1].set_ylabel('Amplitude')
for ax in axes:
    ax.set_xlabel('Amostras')
plt.tight_layout()
```
